```python
df['month'] = df['datetime'].dt.month


df_summer = df[df['month'].isin([5, 6, 7, 8])]   # May–Aug
df_winter = df[df['month'].isin([11, 12, 1, 2])] # Nov–Feb


def create_temp_plot(data, season_name, color1='orange', color2='skyblue'):
    fig = go.Figure()

    fig.add_trace(go.Scatter(
        x=data['datetime'],
        y=data['temp'],
        mode='lines',
        name='Calf Temp (°C)',
        line=dict(color=color1, width=2)
    ))

    fig.add_trace(go.Scatter(
        x=data['datetime'],
        y=data['airTemp'],
        mode='lines',
        name='Outside Air Temp (°C)',
        line=dict(color=color2, width=2, dash='dot')
    ))

    fig.update_layout(
        title=f'Calf vs Outside Temperature — {season_name}',
        xaxis_title='Date & Time',
        yaxis_title='Temperature (°C)',
        legend_title='Temperature Type',
        template='plotly_dark',
        hovermode='x unified',
        xaxis=dict(showgrid=False),
        yaxis=dict(showgrid=True, gridcolor='gray'),
        title_font=dict(size=20, family='Arial', color='white')
    )
    return fig


fig_summer = create_temp_plot(df_summer, "Peak Summer (May–Aug)")
fig_winter = create_temp_plot(df_winter, "Peak Winter (Nov–Feb)")

fig_summer.show()
fig_winter.show()
```

```python
df['month'] = df['datetime'].dt.month


df_summer = df[df['month'].isin([5, 6, 7, 8])]
df_winter = df[df['month'].isin([11, 12, 1, 2])]


TEMP_THRESHOLD = 39.8
THI_THRESHOLD = 74  

def create_temp_plot(data, season_name, color1='orange', color2='skyblue', color3='lightgreen'):
    fig = go.Figure()

    fig.add_trace(go.Scatter(
        x=data['datetime'],
        y=data['temp'],
        mode='lines',
        name='Calf Temp (°C)',
        line=dict(color=color1, width=2)
    ))

    
    fig.add_trace(go.Scatter(
        x=data['datetime'],
        y=data['airTemp'],
        mode='lines',
        name='Outside Air Temp (°C)',
        line=dict(color=color2, width=2, dash='dot')
    ))

    fig.add_trace(go.Scatter(
        x=data['datetime'],
        y=data['humidity'],
        mode='lines',
        name='Humidity / THI',
        yaxis='y2',
        line=dict(color=color3, width=2, dash='dash')
    ))

    
    stress_points = data[(data['temp'] > TEMP_THRESHOLD) & (data['humidity'] > THI_THRESHOLD)]
    fig.add_trace(go.Scatter(
        x=stress_points['datetime'],
        y=stress_points['temp'],
        mode='markers',
        name='⚠️ Heat Stress',
        marker=dict(color='red', size=8, symbol='circle'),
        hovertext=[f"THI={h:.1f}" for h in stress_points['humidity']]
    ))

    
    fig.update_layout(
        title=f'Calf Temp, Air Temp & Humidity (THI) — {season_name}',
        xaxis_title='Date & Time',
        yaxis_title='Temperature (°C)',
        legend_title='Metrics',
        template='plotly_dark',
        hovermode='x unified',
        xaxis=dict(showgrid=False),
        yaxis=dict(showgrid=True, gridcolor='gray'),

    
        yaxis2=dict(
            title='Humidity / THI',
            overlaying='y',
            side='right',
            showgrid=False
        ),
        title_font=dict(size=20, family='Arial', color='white')
    )

    return fig


fig_summer = create_temp_plot(df_summer, "Peak Summer (May–Aug)")
fig_winter = create_temp_plot(df_winter, "Peak Winter (Nov–Feb)")


fig_summer.show()
fig_winter.show()
```

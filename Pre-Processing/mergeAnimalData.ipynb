```python
import pandas as pd
import glob
import os
import re
```

```python
input_folder = "C:/Users/rashe/Desktop/Cattle Scan/Farm Data/Farm-id-68"
```

```python
files = glob.glob(os.path.join(input_folder, "cleaned-animalId-*.csv"))
```

```python
dfs = []

for file in files:
    df = pd.read_csv(file)
    # Extract calf ID for sorting later
    match = re.search(r'animalId-(\d+)', file)
    if match:
        df['id'] = int(match.group(1))
    dfs.append(df)
```

```python
merged_df = pd.concat(dfs, ignore_index=True)
```

```python
merged_df = merged_df.sort_values(by='id')


merged_df.to_csv("animal-data-farm-id-68.csv", index=False) # changge id here

print(f"Successfully merged {len(files)} files into animal-data-farm-id-68.csv") # changge id here
```
